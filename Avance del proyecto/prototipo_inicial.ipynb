```python
planes = {"1": "Básico", "2": "Intermedio", "3": "Premium"}
precios = {"Básico": 299.0, "Intermedio": 399.0, "Premium": 499.0}

socios = {}
siguiente_numero = 1
intentos_acceso = 0
accesos_autorizados = 0


def pedir_numero_socio():
    """Solicita un numero entero positivo sin cerrar el programa por error."""
    while True:
        try:
            numero = int(input("Numero de socio: "))
            if numero > 0:
                return numero
            else:
                print("Error: ingresa un numero mayor que cero.")
        except ValueError:
            print("Error: ingresa un numero entero valido.")


while True:
    print("\n========== GYM NOVA ==========")
    print("1. Ver planes y precios")
    print("2. Registrar socio")
    print("3. Consultar membresia")
    print("4. Activar o reactivar membresia")
    print("5. Cancelar membresia")
    print("6. Validar acceso")
    print("7. Ver socios y estadisticas")
    print("8. Salir")

    opcion = input("Selecciona una opcion: ").strip()

    if opcion == "1":
        print("\nPLANES MENSUALES")
        for clave, plan in planes.items():
            print(clave + ". " + plan + ": $" + str(precios[plan]))

    elif opcion == "2":
        print("\nREGISTRO DE SOCIO")
        nombre = input("Nombre del socio: ").strip()
        while nombre == "":
            print("El nombre no puede quedar vacio.")
            nombre = input("Nombre del socio: ").strip()

        print("Planes: 1. Básico ($299) | 2. Intermedio ($399) | 3. Premium ($499)")
        plan_opcion = input("Elige un plan (1, 2 o 3): ").strip()
        while plan_opcion not in planes:
            print("Plan no valido. Selecciona 1, 2 o 3.")
            plan_opcion = input("Elige un plan (1, 2 o 3): ").strip()

        plan = planes[plan_opcion]
        precio = precios[plan]

        socios[siguiente_numero] = {
            "nombre": nombre,
            "plan": plan,
            "precio": precio,
            "estado": "activa",
            "pago_confirmado": True
        }
        print("Socio registrado correctamente.")
        print("Numero asignado:", siguiente_numero)
        print("Plan:", plan, "| Mensualidad: $", precio)
        print("Membresia activa (pago inicial supuesto como confirmado).")
        siguiente_numero += 1

    elif opcion == "3":
        print("\nCONSULTAR MEMBRESIA")
        numero = pedir_numero_socio()
        if numero not in socios:
            print("No se encontro un socio con ese numero.")
        else:
            socio = socios[numero]
            print("Nombre:", socio["nombre"])
            print("Plan:", socio["plan"])
            print("Precio mensual: $", socio["precio"])
            print("Estado:", socio["estado"])
            if socio["pago_confirmado"]:
                print("Pago confirmado: Sí")
            else:
                print("Pago confirmado: No")

    elif opcion == "4":
        print("\nACTIVAR O REACTIVAR MEMBRESIA")
        numero = pedir_numero_socio()
        if numero not in socios:
            print("No se encontro un socio con ese numero.")
        elif socios[numero]["estado"] == "activa" and socios[numero]["pago_confirmado"]:
            print("Esta membresia ya se encuentra activa.")
        else:
            respuesta = input("¿El pago fue revisado y aprobado? (s/n): ").strip()
            if respuesta == "s" or respuesta == "S":
                socios[numero]["estado"] = "activa"
                socios[numero]["pago_confirmado"] = True
                print("Membresia activada correctamente.")
            else:
                print("No se activo. Es necesario confirmar el pago.")

    elif opcion == "5":
        print("\nCANCELAR MEMBRESIA")
        numero = pedir_numero_socio()
        if numero not in socios:
            print("No se encontro un socio con ese numero.")
        elif socios[numero]["estado"] == "cancelada":
            print("La membresia ya estaba cancelada.")
        else:
            socios[numero]["estado"] = "cancelada"
            socios[numero]["pago_confirmado"] = False
            print("Membresia cancelada. El acceso queda bloqueado.")

    elif opcion == "6":
        print("\nVALIDAR ACCESO")
        numero = pedir_numero_socio()
        intentos_acceso += 1
        if numero not in socios:
            print("ACCESO DENEGADO: socio no encontrado.")
        else:
            socio = socios[numero]
            puede_entrar = socio["estado"] == "activa" and socio["pago_confirmado"]
            if puede_entrar:
                accesos_autorizados += 1
                print("ACCESO AUTORIZADO. Bienvenido/a,", socio["nombre"])
            else:
                print("ACCESO DENEGADO: membresia cancelada o pago no confirmado.")

    elif opcion == "7":
        print("\nLISTA DE SOCIOS")
        if not socios:
            print("Todavia no hay socios registrados.")
        else:
            for numero, socio in socios.items():
                print(numero, "-", socio["nombre"], "|", socio["plan"], "|", socio["estado"])

        total = len(socios)
        activas = 0
        ingresos_teoricos = 0.0
        for socio in socios.values():
            if socio["estado"] == "activa" and socio["pago_confirmado"]:
                activas += 1
                ingresos_teoricos += socio["precio"]

        canceladas = total - activas
        if total > 0:
            porcentaje_activas = activas * 100 / total
        else:
            porcentaje_activas = 0
        print("\nESTADISTICAS")
        print("Total de socios:", total)
        print("Membresias activas:", activas)
        print("Membresias canceladas/inactivas:", canceladas)
        print("Porcentaje de membresias activas:", round(porcentaje_activas, 1), "%")
        print("Ingreso mensual teorico: $", round(ingresos_teoricos, 2))
        print("Intentos de acceso:", intentos_acceso)
        print("Accesos autorizados:", accesos_autorizados)

    elif opcion == "8":
        print("Gracias por usar Gym Nova. Programa finalizado.")
        break

    else:
        print("Opcion no valida. Elige una opcion del 1 al 8.")
```
