```python
# Ejercicio 7: Número par o impar

numero = int(input("Ingresa un número entero: "))

if numero % 2 == 0:
    print(f"{numero} es un numero par")
else:
    print(f"{numero} es un numero impar")
```

```python
# Ejercicio 8: El mayor de tres numeros 

numero1 = int(input("Ingresa el primer numero:"))
numero2 = int(input("Ingresa el segundo numero:"))
numero3 = int(input("Ingresa el tercer numero:"))

if numero1 >= numero2 and numero1 >= numero3:
    print(f"El mayor es: {numero1}")
elif numero2 >= numero1 and numero2 >= numero3:
    print(f"El mayor es: {numero2}")
else:
    print(f"El mayor es: {numero3}")
```

```python
# Ejercicio 10: Tipo de triangulo por sus lados

lado1 = float(input("Ingresa la longitud del primer lado: "))
lado2 = float(input("Ingresa la longitud del segundo lado: "))
lado3 = float(input("Ingresa la longitud del tercer lado: "))

if lado1 == lado2 == lado3:
    print("El triangulo es equilatero")
elif lado1 == lado2 or lado1 == lado3 or lado2 == lado3:
    print("El triangulo es isosceles")
else:
    print("El triangulo es escaleno")
```

```python
# Ejercicio 11: Suma con centinela 0

suma = 0
numero = int(input("Ingresa un numero (0 para terminar): "))

while numero != 0:
    suma += numero
    numero = int(input("Ingresa un numero (0 para terminar): "))

print(f"La suma es: {suma}")
```

```python
# Ejercicio 12: Contar numeros pares

numero = int(input("Ingresa un número: "))
pares = 0

while numero >= 0:
    if numero % 2 == 0:
        pares = pares + 1
    numero = int(input("Ingresa un número: "))

print("Cantidad de números pares:", pares)
```

```python
# Ejercicio 13: Promedio con centinela -1

calificacion = float(input("Ingresa una calificacion (-1 para terminar): "))
suma = 0
contador = 0

while calificacion != -1:
    suma = suma + calificacion
    contador = contador + 1
    calificacion = float(input("Ingresa una calificacion (-1 para terminar): "))

if contador > 0:
    promedio = suma / contador
    print("El promedio es:", promedio)
else:
    print("No se ingresaron calificaciones.")
```

```python
# Ejercicio 16: Suma de los primeros n numeros naturales
n = int(input("Ingresa un numero natural: "))
suma = 0
i = 1

while i <= n:
    suma = suma + i
    i = i + 1

print("La suma de los primeros", n, "numeros naturales es:", suma)
```

```python
# Ejercicio 17: Factorial de un número
n = int(input("Ingresa un número natural: "))
factorial = 1
i = 1

while i <= n:
    factorial = factorial * i
    i = i + 1

print("El factorial de", n, "es:", factorial)
```

```python
# Ejercicio 18: Números pares del 1 al n

n = int(input("Ingresa un número natural: "))
i = 1

while i <= n:
    if i % 2 == 0:
        print(i, end=" ")
    i = i + 1 
```

```python
# Ejercicio 19: Promedio de N calificaciones con for
n = int(input("Ingresa el número de calificaciones: "))
suma = 0

for i in range(n):
    calificacion = float(input("Ingresa la calificación: "))
    suma = suma + calificacion

promedio = suma / n
print("El promedio es:", promedio)
```

```python
# Ejercicio 23: Promedio con validacion de rango

calificacion = float(input("Ingresa una calificacion (0-100): "))
suma = 0
contador = 0

while calificacion >= 0 and calificacion <= 100:
    suma = suma + calificacion
    contador = contador + 1
    calificacion = float(input("Ingresa una calificacion (0-100): "))

if contador > 0:
    promedio = suma / contador
    print("El promedio es:", promedio)
else:
    print("No se ingresaron calificaciones válidas.")
```

```python
# Ejercicio: Promedio con validación de rango

suma = 0
contador = 0

calificacion = float(input("Ingresa una calificacion (0-10) o -1 para terminar: "))

while calificacion != -1:
    if calificacion >= 0 and calificacion <= 10:
        suma = suma + calificacion
        contador = contador + 1
    else:
        print("Error: calificación fuera de rango (0-10)")

    calificacion = float(input("Ingresa una calificacion (0-10) o -1 para terminar: "))

if contador > 0:
    promedio = suma / contador
    print("El promedio es:", promedio)
else:
    print("No se ingresaron calificaciones válidas.")
```
