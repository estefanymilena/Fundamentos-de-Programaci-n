```python
# Taquilla del Museo de Antropología e Historia

# Costos y descuentos
infantil = 0
juvenil = 30
general = 45

# Descuentos
descuentomayor = 0.12
descuentoestudiante = 0.10 

# Contadores
boletos = 0
caja = 0
ahorros = 0

print("--------- TAQUILLA DEL MUSEO ---------")

while True:
    dato = input("\nEdad del visitante (o escribe fin para corte de la caja): ")

    if dato == "fin":
        break

    if not dato:
        print("Edad no válida")
        continue

    anios = int(dato)

    if anios >= 110:
        print("Edad no válida")
        continue

    # Costo del boleto
    rebaja = 0
    clase = "No aplica"

    if anios < 3:
        cuota = infantil 
    elif anios < 18:
        cuota = juvenil
    else:
        cuota = general

        opcion = input("Tipo de visitante (1. Estudiante / 2. Profesor / 3.Ninguno): ")

        while opcion not in ("1", "2", "3"):
            print("Opción no válida")
            opcion = input("Tipo de visitante (1. Estudiante / 2. Profesor / 3.Ninguno): ")

        if opcion == "1":
            clase = "Estudiante"
            
        elif opcion == "2":
            clase = "Profesor"
            
        else:
            clase = "Ninguno"

        # Descuentos
        if anios >= 60:
            rebaja = cuota * descuentomayor
        elif opcion == "1" or opcion == "2":
            rebaja = cuota * descuentoestudiante

    # Calcular lo que paga
    cobro = cuota - rebaja

    boletos += 1
    caja += cobro
    ahorros += rebaja

    print(f"Visitante {boletos} -- Edad: {anios} -- Tipo: {clase} -- Base: ${cuota:.2f} -- Descuento: ${rebaja:.2f} -- Paga: ${cobro:.2f}")

# Corte final
print("\n--------- CORTE DE CAJA ---------")
print(f"Cantidad de visitantes: {boletos}")
print(f"Descuentos en pesos: ${ahorros:.2f}")
print(f"Total: ${caja:.2f}")
```
