```python
def es_mayor(edad):
    return True if edad >= 18 else False

edades = [15, 22, 17, 30, 19, 12]
personas = ["Sofa", "Nicolette", "Milena", "Fernanda", "Josue", "Fanny"]

for i in range(len(edades)):
    if es_mayor(edades[i]):
        print(personas[i], edades[i], "años: pasa")
    else:
        print(personas[i], edades[i], "años: no pasa") 
```

```python
# R3 Promedio por alumno
def promedio(fila):
    promedio = sum(fila) / len(fila)
    return promedio

calificaciones = [8, 9, 10], [6, 7, 8]
print("Alumno 1:", promedio(calificaciones[0]))
print("Alumno 2:", promedio(calificaciones[1]))
```

```python
def promedio(fila):
    promedio = sum(fila) / len(fila)
    return promedio
calificaciones = [8, 9, 10], [6, 7, 8]

for i , fila in enumerate(calificaciones, start=1):
    prom = promedio(fila)
    print(f"alumno{i}: {prom:2f}")
```
