# Descripción del programa.
Este programa hace una tabla de Pitágoras de 10 por 10 utilizando una matriz y la ejecuta de forma ordenada, tambien permite consultar el resultado de una multiplicación mediante la búsqueda de valores en la matriz, mientras que los factores ingresados estén entre 1 y 10.

```python
# Parte A - Construir la matriz
limite = 10
matriz = []

for numero_fila in range(1, limite + 1):
    nueva_fila = []
    
    for numero_columna in range(1, limite + 1):
        nueva_fila.append(numero_fila * numero_columna)
    
    matriz.append(nueva_fila)

# Parte B - Mostrarla: funcion que no regresa valor
def imprimir_tabla_de_pitagoras(matriz):
    print("\t", end="")
    
    for encabezado in range(1, limite + 1):
        print(encabezado, end="\t")
    
    print()
    
    for posicion in range(len(matriz)):
        print(posicion + 1, end="\t")
        
        for valor in matriz[posicion]:
            print(valor, end="\t")
        
        print()

# Parte C - Consultarla: funcion que si regresa valor
def consultar_tabla(matriz, renglon, columna):
    return matriz[renglon - 1][columna - 1]

imprimir_tabla_de_pitagoras(matriz)

# Parte D - Programa principal
numero_fila = int(input("Escribe el primer factor: "))

while numero_fila < 1 or numero_fila > 10:
    print("Error: el numero debe estar entre 1 y 10.")
    numero_fila = int(input("Escribe nuevamente el primer factor: "))


numero_columna = int(input("Escribe el segundo factor: "))

while numero_columna < 1 or numero_columna > 10:
    print("Error: el numero debe estar entre 1 y 10.")
    numero_columna = int(input("Escribe nuevamente el segundo factor: "))


producto = consultar_tabla(matriz, numero_fila, numero_columna)

print(numero_fila, "x", numero_columna, "=", producto)

```
